#### Explore gold.fact_product_pairs

```python
from warehouse import Warehouse

with Warehouse() as wh:
    df_fact_product_pairs = wh.read_table(name="fact_product_pairs", schema="gold")
df_fact_product_pairs.info()
df_fact_product_pairs.head()
```

```python
# beyond info: confidence must sit in [0,1] and how many pairs beat chance
conf = df_fact_product_pairs[["confidence_a_to_b", "confidence_b_to_a"]]
assert ((conf >= 0) & (conf <= 1)).all().all()
print("lift > 1:", (df_fact_product_pairs["lift"] > 1).sum(), "of", len(df_fact_product_pairs))
print("lift range:", round(df_fact_product_pairs["lift"].min(), 3), "to", round(df_fact_product_pairs["lift"].max(), 3))
```
